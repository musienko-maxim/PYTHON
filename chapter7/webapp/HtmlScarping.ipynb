```python
import gazpacho
import gazpacho.get
import json
import swimclub

RECORDS = (0,1,3,4)
COURSES = ("LC Men","LC Women", "SC Men", "SC Women")


list(zip(RECORDS, COURSES))

URL = "https://en.wikipedia.org/wiki/List_of_world_records_in_swimming"
html = gazpacho.get(URL)
soup = gazpacho.Soup(html)
tables = soup.find("table")
# for n, table in  enumerate(tables):
#     rows = table.find("tr", mode="all")
#     lastRow = rows[-1]
#     columns = lastRow.find("td", mode="all")
#     print(f"Table {n} has {len(rows)} rows, the last row has {len(columns)} column")


# table = tables[0]
# rows = table.find("tr", mode="all")
# for row in rows[1:]:
#     columns = row.find("td",mode="all")
#     event = columns[0].text
#     time = columns[1].text
#     print(f" {event} -> time {time}")


conversion = {
    "Free":"freestyle",
    "Back":"backstore",
    "Breast":"breaststoke",
    "Fly":"butterfly",
    "IM":"individual medley"
}
event = "Mike-15-100m-Fly.txt"
*_, distance, stroke = event.removesuffix(".txt").split("-")
lookup = f"{distance} {conversion[stroke]}"
records = {}        
for table, course in zip (RECORDS, COURSES):
    records[course]={}
    #print(f"{course}:")
    for row in tables[table].find("tr", mode="all")[1:]:
        columns = row.find("td",mode="all")
        event = columns[0].text
        time = columns[1].text
        if "relay" not in event:
            records[course][event] = time

#print(swimclub.event_lookup("Darius-13-100m-Back.txt"))
def pr_bar_chart(fn):
    times = records[course][swimclub.event_lookup(fn)]
    return times

with open("records.json", "w") as jf:
     json.dump(records, jf)

with open("records.json") as jf:
    records = json.load(jf)

for course in COURSES:
    print(records[course][swimclub.event_lookup("Darius-13-100m-Fly.txt")])

```
